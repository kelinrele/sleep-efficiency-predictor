```python
df_cleaned = df.dropna()
```

```python
df_cleaned.shape
```

```python

df_cleaned['date'] = pd.to_datetime(df_cleaned['date'])

df_cleaned = df_cleaned.sort_values(by=['user_id', 'date']).reset_index(drop=True)


category_cols = ['gender', 'region', 'workout_type', 'mood']
for col in category_cols:
    print(f"\n--- {col} ---")
    print(df_cleaned[col].value_counts())
```

```python
cols_to_drop = ['region', 'device_model', 'height_cm', 'weight_kg']
df_encoded = df_cleaned.drop(columns=cols_to_drop).copy()

mood_mapping = {'very_bad': 1, 'bad': 2, 'neutral': 3, 'good': 4, 'very_good': 5}
df_encoded['mood_encoded'] = df_encoded['mood'].map(mood_mapping)
df_encoded = df_encoded.drop(columns=['mood'])

df_encoded = pd.get_dummies(df_encoded, columns=['gender', 'workout_type'], drop_first=True)

print(f"New column count: {df_encoded.shape[1]}")
print(df_encoded.columns.tolist())
```

```python
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler
y = df_encoded['sleep_efficiency']
X = df_encoded.drop(columns=['sleep_efficiency','user_id','date'])
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
lasso = Lasso(alpha=0.001, random_state=42) 
lasso.fit(X_scaled, y)
coefficents = lasso.coef_
lasso_coefs = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': lasso.coef_
}).sort_values(by='Coefficient', key=abs, ascending=False)

surviving_features = lasso_coefs[lasso_coefs['Coefficient'] != 0]

print("--- Surviving Features from Lasso ---")
print(surviving_features.to_string(index=False))
print(f"\nFeatures reduced from {X.shape[1]} to {surviving_features.shape[0]}")
```

```python
from sklearn.linear_model import LassoCV

lasso_cv = LassoCV(alphas = [0.001, 0.0012, 0.0013, 0.0014, 0.0015, 0.0016, 0.0017, 0.0018, 0.0019, 0.0110], cv=5, random_state=42)

lasso_cv.fit(X_scaled, y)

print(f"The mathematically optimal alpha is: {lasso_cv.alpha_}")
 
```

```python
import numpy as np

mean_mses = np.mean(lasso_cv.mse_path_, axis=1)

error_df = pd.DataFrame({
    'Alpha (Penalty)': lasso_cv.alphas_,
    'Average MSE': mean_mses
})

error_df = error_df.sort_values(by='Alpha (Penalty)')

print("--- Cross-Validation Mean Squared Errors ---")
print(error_df.to_string(index=False))
```

```python
features_to_test = ['steps', 'alcohol_units', 'stress_score', 'sleep_stage_light_pct', 'sleep_stage_deep_pct', 'caffeine_mg', 'workout_type_none', 'screen_time_min', 'workout_type_walk']

correlations = df_encoded[features_to_test + ['sleep_efficiency']].corr()['sleep_efficiency']

print(correlations.sort_values(ascending=False))
```

```python
X_final = df_encoded[['steps','alcohol_units','stress_score','caffeine_mg','workout_type_none','screen_time_min','workout_type_walk']]

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_final, y, test_size=0.2, random_state=42
)
X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)

lasso.fit(X_train, y_train)

```

```python
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error

y_pred = lasso.predict(X_test)

mse = mean_squared_error(y_test, y_pred)

r2 = r2_score(y_test, y_pred)

mse, r2
```

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=100,       
    learning_rate=0.1,     
    max_depth=5,           
    random_state=42
)

xgb_model.fit(X_train, y_train)

xgb_predictions = xgb_model.predict(X_test)

mse = mean_squared_error(y_test, xgb_predictions)

r2 = r2_score(y_test, xgb_predictions)

mse, r2
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.05, 0.1],
    'n_estimators': [100, 200, 300]
}

grid_search = GridSearchCV(
    estimator=XGBRegressor(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring='neg_mean_squared_error',
    n_jobs=-1
)
grid_search.fit(X_train, y_train)
```

```python
print("Winning Parameters:", grid_search.best_params_)

best_xgb = grid_search.best_estimator_

final_predictions = best_xgb.predict(X_test)
final_mse = mean_squared_error(y_test, final_predictions)
final_r2 = r2_score(y_test, final_predictions)

print(f"\nFinal Optimized MSE: {final_mse:.5f}")
print(f"Final Optimized R^2: {final_r2:.4f}")
```

```python
param_grid = {
    'max_depth': [2, 3, 4],               # Pushing below 3
    'learning_rate': [0.03, 0.05, 0.06,0.07],  # Micro-tuning around 0.05
    'n_estimators': [50, 75, 100]         # Pushing below 100
}
grid_search = GridSearchCV(
    estimator=XGBRegressor(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring='neg_mean_squared_error',
    n_jobs=-1
)
grid_search.fit(X_train, y_train)
```
